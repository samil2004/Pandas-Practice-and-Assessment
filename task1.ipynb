
# 1️⃣ Create a DataFrame

Create a DataFrame with columns: Name, Age, and Marks
Include 4 student records with at least one missing value.
(Hint: Use a dictionary and pd.DataFrame())


```python
import pandas as pd
data = {
    'Name': ['samil', 'sabu', 'Rahul', 'sahal'],
    'Age': [21, 22, None, 24],
    'Marks': [85, 92, 78, 95]
}

df = pd.DataFrame(data)
print("1. DataFrame:")
print(df)
```

# 2️⃣ Add a new column

Add a new column "Grade" with letter grades (A, B, C, etc.)

```python
df['Grade']=['A','B','A','C']
print('ADD GRADE')
df
```

Count the number of missing values in each column.
(Hint: df.isnull().sum())

```python
df.isnull().sum()
```

# 5️⃣ Handle missing values

Replace all missing values in 'Age' with the mean age.
Hint: df['Age'].fillna(df['Age'].mean(), inplace=True))

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
print(df['Age'])
```

# 7️⃣ Add new rows to the DataFrame

Add two new rows:
{'Name': 'Kumar', 'Age': 20, 'Marks': 90}
{'Name': 'Rahul', 'Age': 26, 'Marks': 85}

# 1️⃣1️⃣ Aggregate functions

Calculate and display:
 Mean Age
 Minimum Age
 Maximum Age
 Count of Age values

```python
print("Mean Age:", df['Age'].mean())
print("Minimum Age:", df['Age'].min())
print("Maximum Age:", df['Age'].max())
print("Count of Age:", df['Age'].count())
```
