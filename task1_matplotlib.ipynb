# 1️⃣ LINE GRAPH

### Q1: Create a line graph to show the population growth of India from 2000 to 2020.
### years = [2000, 2005, 2010, 2015, 2020]
### population = [1.0, 1.1, 1.2, 1.3, 1.38]
### Add labels for x and y axes
### Add title "Population Growth of India (2000–2020)"
### Display grid lines

```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
x=np.array([2000, 2005, 2010, 2015, 2020])
y=np.array([1.0,1.1,1.2,1.3,1.38])
plt.plot(x,y,marker='v',#o
         markersize=8,
         markerfacecolor='red',
         markeredgecolor='black',
         linestyle='solid',#dotted,dashed,dashdot
         linewidth=2,
         color='black')
plt.xlabel('yeras',fontsize=15,color="black")
plt.ylabel('population',fontsize=15,color="black")
plt.title('Population Growth of India (2000–2020)',fontweight='bold',fontsize=20,color="black")
plt.grid(True)
plt.show()
```

# Q2: Plot two lines on the same graph — sales of Product A and Product B over 6 months.
### months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
### productA = [120, 150, 170, 190, 210, 230]
### productB = [100, 130, 160, 180, 200, 250]
### - Use different colors and line styles
### - Add a legend

```python

import matplotlib
import matplotlib.pyplot as plt
import numpy as np

x1 = np.array(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'])
y1 = np.array([120, 150, 170, 190, 210, 230])
y2 = np.array([100, 130, 160, 180, 200, 250])

plt.plot(x1, y1,
         color='blue',
         linestyle='-',
         marker='o',
         markersize=8,
         linewidth=2,
         label='Product A')

plt.plot(x1, y2,
         color='green',
         linestyle='--',
         marker='v',
         markersize=8,
         linewidth=2,
         label='Product B')

plt.xlabel('Months', fontsize=15, color='black')
plt.ylabel('Sales', fontsize=15, color='black')

plt.title('Monthly Sales of Product A and Product B',
          fontweight='bold',
          fontsize=18,
          color='black')

plt.legend()
plt.grid(True)
plt.show()
```

# Q4: Plot a grouped bar chart comparing male vs female students in each department.
### departments = ['CSE', 'ECE', 'EEE', 'ME']
### male = [40, 35, 30, 25]
### female = [30, 25, 20, 15]
### - Display bars side by side
### - Add a legend

```python

import matplotlib.pyplot as plt
import numpy as np

departments = np.array(['CSE', 'ECE', 'EEE', 'ME'])
male = np.array([40, 35, 30, 25])
female = np.array([30, 25, 20, 15])

x = np.arange(len(departments))
width = 0.35

plt.bar(x - width/2, male, width,
        color='skyblue', label='Male')

plt.bar(x + width/2, female, width,
        color='pink', label='Female')

plt.xlabel('Departments', fontsize=15)
plt.ylabel('Number of Students', fontsize=15)

plt.title('Male vs Female Students by Department',
          fontweight='bold', fontsize=18)

plt.xticks(x, departments)

plt.legend()
plt.show()
```

# Q5: Draw a pie chart showing the market share of mobile brands.
# brands = ['Samsung', 'Apple', 'Xiaomi', 'OnePlus', 'Others']
# share = [30, 25, 20, 15, 10]
# - Display percentage values
# - Explode the slice with the highest share

```python

import matplotlib.pyplot as plt
import numpy as np

brands = np.array(['Samsung', 'Apple', 'Xiaomi', 'OnePlus', 'Others'])
share = np.array([30, 25, 20, 15, 10])

color = ['red', 'blue', 'green', 'orange', 'yellow']
explode = [0.1, 0, 0, 0, 0]

plt.pie(share,
        labels=brands,
        autopct="%1.1f%%",
        colors=color,
        explode=explode)

plt.title("Market Share of Mobile Brands",
          fontsize=20,
          color='red',
          fontweight='bold')

plt.show()
```

# Q6: Create a pie chart representing the distribution of expenses in a household budget.
### categories = ['Rent', 'Food', 'Transport', 'Entertainment', 'Savings']

# Q8: Plot two scatter plots on the same figure for Class A and Class B students’ height and weight.
# height_A = [150, 160, 165, 170, 175]
# weight_A = [50, 55, 60, 65, 70]
# height_B = [152, 158, 168, 172, 178]
# weight_B = [48, 53, 62, 66, 72]
# - Use different colors and markers
# - Add a legend

```python

import matplotlib.pyplot as plt
import numpy as np

height_A = np.array([150, 160, 165, 170, 175])
weight_A = np.array([50, 55, 60, 65, 70])

height_B = np.array([152, 158, 168, 172, 178])
weight_B = np.array([48, 53, 62, 66, 72])


plt.scatter(height_A, weight_A,
            color='blue', marker='o',
            label='Class A')


plt.scatter(height_B, weight_B,
            color='red', marker='^',
            label='Class B')

plt.xlabel('Height (cm)', fontsize=15)
plt.ylabel('Weight (kg)', fontsize=15)

plt.title('Height vs Weight: Class A and Class B',
          fontsize=18, fontweight='bold')

plt.legend()
plt.grid(True)
plt.show()
```

# Q10: Generate a histogram for 100 random numbers normally distributed around mean = 50 and std = 10 using NumPy.
# - Plot histogram with 10 bins
# - Add grid and color customization

```python

data = np.random.normal(loc=50, scale=10, size=100)


plt.hist(data,
         bins=10,
         color='skyblue',
         edgecolor='black')

plt.title('Histogram of Normally Distributed Random Numbers',
          fontsize=18,
          color='red',
          fontweight='bold')

plt.xlabel('Values', fontsize=15)
plt.ylabel('Frequency', fontsize=15)


plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

# 2️⃣ BAR GRAPH: Total Sales by Country
# ============================================================
# Task: Create a bar chart showing total sales for each country.
# Hint:
#   1. Add a 'Sales' column: df['Sales'] = df['Quantity'] * df['Price']
#   2. Group by 'Country' and sum 'Sales'
#   3. Sort values in descending order
#   4. Use plt.bar() to plot
#   5. Display the sales value above each bar

```python
import matplotlib.pyplot as plt
import pandas as pd


dataset = {
    'Country': ['India', 'USA', 'UK', 'India', 'USA', 'UK'],
    'Quantity': [10, 20, 15, 5, 10, 8],
    'Price': [100, 200, 150, 120, 180, 100]
}

df = pd.DataFrame(dataset)


df['Sales'] = df['Quantity'] * df['Price']


country_sales = df.groupby('Country')['Sales'].sum()


country_sales = country_sales.sort_values(ascending=False)


bars = plt.bar(
    country_sales.index,
    country_sales.values,
    color='skyblue',
    edgecolor='black'
)


plt.bar_label(bars, fmt='%.0f', padding=3)

plt.xlabel('Country', fontsize=15)
plt.ylabel('Total Sales', fontsize=15)

plt.title(
    'Total Sales by Country',
    fontsize=18,
    color='red',
    fontweight='bold'
)


plt.show()
```

# 3️⃣ LINE GRAPH: Daily Sales Trend
# ============================================================
# Task: Plot a line graph showing total sales per day.
# Hint:
#   1. Convert 'OrderDate' to datetime: df['OrderDate'] = pd.to_datetime(df['OrderDate'])
#   2. Group by 'OrderDate' and sum 'Sales'
#   3. Use plt.plot() with markers and line style
#   4. Label X-axis, Y-axis, and add a title

```python
import matplotlib.pyplot as plt
import pandas as pd

dataset = {
    'OrderDate': [
        '2026-10-20', '2026-10-20',
        '2026-10-21', '2026-10-21',
        '2026-10-22', '2026-10-23'
    ],
    'Sales': [1000, 1500, 2000, 1200, 1800, 2500]
}

df = pd.DataFrame(dataset)


df['OrderDate'] = pd.to_datetime(df['OrderDate'])

daily_sales = df.groupby('OrderDate')['Sales'].sum()

plt.plot(
    daily_sales.index,
    daily_sales.values,
    marker='o',
    linestyle='-',
    color='blue',
    linewidth=2,
    markersize=7
)


plt.xlabel('Order Date', fontsize=12)
plt.ylabel('Total Sales', fontsize=12)

plt.title(
    'Daily Sales Trend',
    fontsize=18,
    color='red',
    fontweight='bold'
)

plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

# 4️⃣ LINE GRAPH: Category-wise Average Price
# ============================================================
# Task: Draw a line chart showing the average product price per category.
# Hint:
#   1. Group by 'Category' and calculate mean of 'Price'
#   2. Use plt.plot() with markers, colors, and line style
#   3. Label axes and add grid lines

```python
import matplotlib.pyplot as plt
import pandas as pd


dataset = {
    'Category': ['Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet'],
    'Price': [50000, 20000, 15000, 60000, 25000, 18000]
}

df = pd.DataFrame(dataset)


category_avg_price = df.groupby('Category')['Price'].mean()


plt.plot(
    category_avg_price.index,
    category_avg_price.values,
    marker='o',
    color='green',
    linestyle='-',
    linewidth=2,
    markersize=8
)


plt.xlabel('Category', fontsize=12)
plt.ylabel('Average Price', fontsize=12)

plt.title(
    'Category-wise Average Price',
    fontsize=18,
    color='red',
    fontweight='bold'
)

plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

# 5️⃣ PIE CHART: Order Distribution by Category
# ============================================================
# Task: Create a pie chart showing the percentage of total orders for each category.
# Hint:
#   1. Use df['Category'].value_counts() to count orders
#   2. Use plt.pie() with autopct='%1.1f%%' and startangle=90
#   3. Add a title

```python
import matplotlib.pyplot as plt
import pandas as pd

dataset = {
    'Category': [
        'Laptop', 'Mobile', 'Tablet',
        'Laptop', 'Mobile', 'Laptop',
        'Tablet', 'Mobile', 'Laptop', 'Tablet'
    ]
}

df = pd.DataFrame(dataset)


category_orders = df['Category'].value_counts()

plt.pie(
    category_orders.values,
    labels=category_orders.index,
    autopct='%1.1f%%',
    startangle=90
)


plt.title(
    'Order Distribution by Category',
    fontsize=18,
    color='red',
    fontweight='bold'
)

plt.show()
```

# 6️⃣ PIE CHART: Country-wise Share of Total Sales
# ============================================================
# Task: Plot a pie chart showing the share of total sales by country.
# Hint:
#   1. Group by 'Country' and sum 'Sales'
#   2. Use plt.pie() with labels, autopct='%1.1f%%', colors, and shadow=True
#   3. Add a title

```python
import matplotlib.pyplot as plt
import pandas as pd


dataset = {
    'Country': ['India', 'USA', 'UK', 'India', 'USA', 'UK'],
    'Sales': [1000, 2000, 1500, 600, 1800, 1000]
}

df = pd.DataFrame(dataset)


country_sales = df.groupby('Country')['Sales'].sum()


plt.pie(
    country_sales.values,
    labels=country_sales.index,
    autopct='%1.1f%%',
    colors=['skyblue', 'orange', 'lightgreen'],
    shadow=True

)


plt.title(
    'Country-wise Share of Total Sales',
    fontsize=18,
    color='red',
    fontweight='bold'
)

plt.show()
```
